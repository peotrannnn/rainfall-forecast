```python
import os
import xarray as xr
import pandas as pd
import zipfile
from pathlib import Path
from tqdm import tqdm
```

```python
cwd = Path.cwd()
possible_paths = [
    cwd / "Data.zip",
    cwd / "Data" / "Data.zip",
    cwd.parent / "Data" / "Data.zip",
]
ZIP_PATH = next((p for p in possible_paths if p.exists()), None)
if ZIP_PATH is None:
    raise FileNotFoundError(
        "Data.zip not found. Checked paths: " + ", ".join(str(p) for p in possible_paths)
    )
OUTPUT_CSV = ZIP_PATH.parent / "Data.csv"
```

```python
first = True

with zipfile.ZipFile(ZIP_PATH, 'r') as z:
    nc_files = [f for f in z.namelist() if f.endswith('.nc')]
    
    print(f"Total .nc files: {len(nc_files)}")
    print(f"Using ZIP_PATH: {ZIP_PATH}")

    for nc_file in tqdm(nc_files, desc="Processing files"):
        with z.open(nc_file) as f:
            ds = xr.open_dataset(f, engine="h5netcdf")
            df = ds.to_dataframe().reset_index()

            df.to_csv(OUTPUT_CSV, mode='a', index=False, header=first)
            first = False

print("Done! Saved to", OUTPUT_CSV)
```
